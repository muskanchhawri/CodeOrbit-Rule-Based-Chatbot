## Project Description

This project is a simple Rule-Based Chatbot developed using Python. It interacts with users, understands basic text inputs, and provides predefined responses based on user messages.

The chatbot uses conditional statements and loops to handle conversations and provides a simple interactive experience.

## Features

- Responds to greetings like Hi, Hello, and Hey.
- Answers basic questions.
- Tells the current date.
- Tells the current time.
- Handles unknown questions with a fallback response.
- Continues conversations until the user types "bye".
- Provides a friendly welcome and goodbye message.

## Technologies Used

- Python
- Jupyter Notebook
- datetime module
- Conditional Statements (if, elif, else)
- While Loop
- User Input and Output

## How to Run

1. Download or clone this repository.
2. Open Jupyter Notebook.
3. Open `Rule_Based_Chatbot.ipynb`.
4. Run the notebook cells.
5. Enter your messages in the input field.
6. Type `bye` to exit the chatbot.
